```python
import random
from transformers import AutoTokenizer, AutoModelForCausalLM, Trainer, TrainingArguments

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("facebook/opt-350m")
model = AutoModelForCausalLM.from_pretrained("facebook/opt-350m")

# Define a sample dataset (replace with your dataset)
from datasets import load_dataset
dataset = load_dataset("wikitext", "wikitext-2-raw-v1")
tokenized_dataset = dataset.map(
    lambda examples: tokenizer(examples['text'], truncation=True, padding="max_length", max_length=512),
    batched=True
)

# Define ranges for hyperparameters
learning_rate_range = [1e-5, 5e-5]
batch_size_options = [8, 16, 32]
epoch_options = [3, 5, 7]

# Perform random search
n_trials = 5  # Number of random trials
for trial in range(n_trials):
    # Randomly select hyperparameters
    learning_rate = random.uniform(learning_rate_range[0], learning_rate_range[1])
    batch_size = random.choice(batch_size_options)
    epochs = random.choice(epoch_options)

    print(f"Trial {trial + 1}: LR={learning_rate}, Batch Size={batch_size}, Epochs={epochs}")

    training_args = TrainingArguments(
        output_dir=f"./results_trial{trial + 1}",
        evaluation_strategy="epoch",
        learning_rate=learning_rate,
        per_device_train_batch_size=batch_size,
        num_train_epochs=epochs,
        save_strategy="epoch",
        logging_dir=f"./logs_trial{trial + 1}",
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_dataset["train"],
        eval_dataset=tokenized_dataset["validation"],
    )

    # Train the model
    trainer.train()

    # Save the model for this trial
    trainer.save_model(f"./model_trial{trial + 1}")

print("Random search completed!")
```
